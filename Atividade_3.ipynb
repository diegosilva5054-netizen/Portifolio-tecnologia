```python
# Importando as bibliotecas
import sqlite3 as sql
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Criando o banco de dados
conexao = sql.connect('dados_vendas.db')
# É para executar comandos em SQL
cursor = conexao.cursor()

# Criando uma tabela
cursor.execute("""
    CREATE TABLE IF NOT EXISTS vendas1 (
        id_vendas INTEGER PRIMARY KEY AUTOINCREMENT,
        data_venda 'DATE',
        produto TEXT,
        categoria TEXT,
        valor_venda REAL
    )
""")

# Limpa tudo que está na tabela, sem deixar repetição.
cursor.execute('DELETE FROM vendas1')

# Inserindo os dados
cursor.execute('''
INSERT INTO vendas1(data_venda, produto, categoria, valor_venda) VALUES
('2023-01-01', 'Produto A', 'Eletrônicos', 1500.00),
('2023-01-05', 'Produto B', 'Roupas', 350.00),
('2023-02-10', 'Produto C', 'Eletrônicos', 1200.00),
('2023-03-15', 'Produto D', 'Livros', 200.00),
('2023-03-20', 'Produto E', 'Eletrônicos', 800.00),
('2023-04-02', 'Produto F', 'Roupas', 400.00),
('2023-05-05', 'Produto G', 'Livros', 150.00),
('2023-06-10', 'Produto H', 'Eletrônicos', 1000.00),
('2023-07-20', 'Produto I', 'Roupas', 600.00),
('2023-08-25', 'Produto J', 'Eletrônicos', 700.00),
('2023-09-30', 'Produto K', 'Livros', 300.00),
('2023-10-05', 'Produto L', 'Roupas', 450.00),
('2023-11-15', 'Produto M', 'Eletrônicos', 900.00),
('2023-12-20', 'Produto N', 'Livros', 250.00);
''')
# Salva as alterações
conexao.commit()

# Exibindo a tabela inteira em um dataframe
df = pd.read_sql_query('SELECT * FROM vendas1', conexao)
print(df)
conexao.close()

# Mudando o tipo data_texto para tipo datetime
df['data_venda'] = pd.to_datetime(df['data_venda'])
# Cria uma coluna com o numero do mês
df['mes_venda'] = df['data_venda'].dt.month

# Agrupa as vendas por categoria e faz a soma por venda
soma_por_categoria = df.groupby('categoria')['valor_venda'].sum()
print('A soma por categoria é:')
print(soma_por_categoria)

# Agrupa os meses das vendas junto com o valor e calcula o total por mês
total_mes = df.groupby('mes_venda')['valor_venda'].sum()
print('O total por mês é:')
print(total_mes)

# Procura a maior valor da coluna valor_vendas
posicao = df['valor_venda'].idxmax()
produto_top = df.loc[posicao]
print('O produto mais vendido é:')
print(produto_top)

# Agrupa as vendas por categoria e cacula a média, round é pra mostrar apenas 2 casas dcimais
ticket_medio = df.groupby('categoria')['valor_venda'].mean().round(2)
print('O ticket médio é:')
print(ticket_medio)

# Define o estilo visual dos graficos
sns.set(style='whitegrid')
meses = ['Janeiro', 'Fevereiro', 'Março', 'Abril', 'Maio', 'Junho', 'Julho', 'Agosto', 'Setembro', 'Outubro', 'Novembro', 'Dezembro']

# Define o tamanho que vai ficar
fig, ax = plt.subplots(2, 2, figsize=(12, 6))

# Adicionando titulo e tamanho da fonte
fig.suptitle('Análise de Vendas', fontsize = 10, fontweight = 'bold', y=1.01)

#Grafico 1 = total por categoria
sns.barplot(x=soma_por_categoria.index, y=soma_por_categoria.values, ax=ax[0, 0], color='steelblue')
ax[0, 0].set_title('Total por Categoria')

#Grafico 2 = total por mês
sns.lineplot(x=total_mes.index, y=total_mes.values, ax=ax[0, 1], marker='s', color='gray')
ax[0, 1].set_title('Vendas por Mês')

#Grafico 3 = Porcentagem por categoria
ax[1, 0].pie(soma_por_categoria.values, labels=soma_por_categoria.index, autopct='%1.1f%%',)
ax[1, 0].set_title('Porcentagem por Categoria')

#Grafico 4 = ticket médio
sns.barplot(x=ticket_medio.values, y=ticket_medio.index, ax=ax[1, 1], color='mediumseagreen')
ax[1, 1].set_title('Ticket Médio por Categoria')

# Arruma o espaçamento entre os graficos
plt.tight_layout()

# Exibe todos os graficos na tela
plt.show()
```
